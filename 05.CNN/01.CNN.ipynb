```python
import torch
from torch import nn
```

```python
layer = nn.Conv2d(in_channels=1, out_channels=2, kernel_size=3)        # 디폴트값 : stride = 1(몇칸씩 움직일건지), padding=0(테두리 칸 수) /in_channels=1, out_channels=2, kernel_size=3은 무조건 입력(이름 생략하고 숫자만 입력 가능)
layer(torch.randn(32,1,5,5)).shape
```

```python
layer = nn.Conv2d(3, 5, 3)
layer(torch.randn(32,3,5,5)).shape

print(layer.weight.shape)       # 가중치의 크기와 모양[특징(출력), 채널(입력), 행, 열]
```

```python
conv1 = nn.Conv2d(1, 8, 6, stride=2)
x = torch.randn(32,1,28,28)
print(conv1(x).shape)

conv2 = nn.Conv2d(8, 16, 3,padding=1)
print(conv2(conv1(x)).shape)

maxpool = nn.MaxPool2d(kernel_size=2, stride=2)
print(maxpool(conv2(conv1(x))).shape)
```

```python
maxpool = nn.MaxPool2d(2)       # kernel_size =2, stride=2      (2x2 영역에서 최댓값을 계산하고, 가로와 세로 크기를 절반으로 줄이는 역할)
x = torch.randn(1,6,6)
print(x.shape)
print(maxpool(x).shape)
print(maxpool(torch.randn(32,3,6,6)).shape)
```

```python
maxpool = nn.AvgPool2d(2)       #  kernel_size =2, stride=2     (2×2 영역의 평균을 계산하고, 가로와 세로 크기를 절반으로 줄이는 역할) -> maxpool과 shape은 같지만 데이터 값이 다름
x = torch.randn(1,6,6)
print(x)
print(maxpool(x))
print(maxpool(torch.randn(32,3,6,6)).shape)
```

```python
class CNN(nn.Module):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.conv1 =nn.Conv2d(1,8,6, stride=2)
        self.conv2 = nn.Conv2d(8,16,3,padding=1)
        self.Maxpool = nn.MaxPool2d(2)
        self.fc = nn.Linear(16*6*6,10)

    def forward(self, x):
        x = self.conv1(x)
        print('conv1',x.shape)
        x = self.conv2(x)
        print('conv2',x.shape)
        x = self.Maxpool(x)
        print('maxpool',x.shape)
        x = torch.flatten(x, start_dim=1) 
        x = self.fc(x)
        return x

x = torch.randn(32, 1, 28,28)
model = CNN()
print(model(x).shape)       # x의 이미지 개수, linear출력개수 -> (32, 10)

# (input+(2*padding)-kernel)/stride + 1= 결과 값(maxpool : (행,열)/2)
# conv1(x) -> (28+0-6)/2 + 1 = 12  (32,8,12,12)
# conv2(x) -> (12+2-3)/1 + 1 = 12 (32,16,12,12)
# maxpool(x) -> (12, 12)/2 = 6 (32, 16, 6, 6)
```
