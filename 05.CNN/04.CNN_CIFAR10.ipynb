```python
from multiclass_functions import *
import torch
from torch import nn, optim
from torchvision import datasets, transforms
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader
import os
os.makedirs('results', exist_ok=True)
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print(DEVICE)
```

```python
BATCH_SIZE = 32
LR = 1e-3
EPOCH = 5
criterion = nn.CrossEntropyLoss()
new_model_train = True

model_type = 'CNN'
dataset = 'CIFAR10'

save_model_path = f'results/{model_type}_{dataset}.pt'
```

```python
transform = transforms.ToTensor()
train_DS = datasets.CIFAR10(root='data', train=True, download = True, transform= transform)
test_DS = datasets.CIFAR10(root='data', train=False, download = True, transform= transform)
train_DL = DataLoader(train_DS, batch_size=BATCH_SIZE, shuffle = True)
test_DL = DataLoader(test_DS, batch_size=BATCH_SIZE, shuffle = True)
```

```python
print(test_DS.classes)
print(test_DS.class_to_idx)

x_batch, y_batch = next(iter(train_DL))
print(x_batch.shape)

plt.imshow(x_batch[0].permute(1,2,0))
print(test_DS.classes[y_batch[0]])
```

```python
class CNN(nn.Module):                                               # x_batch.shape -> (32, 3, 32, 32)
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.conv1 = nn.Sequential(nn.Conv2d(3, 8, 3, padding = 1),     # (32,8,32,32)
                                    nn.BatchNorm2d(8),
                                    nn.ReLU(),
                                    )
        self.Maxpool1 = nn.MaxPool2d(2)                                 # (32,8,16,16)
        self.conv2 = nn.Sequential(nn.Conv2d(8, 16, 3, padding=1),      # (32,16,8,8)
                                    nn.BatchNorm2d(16),
                                    nn.ReLU()
                                    )
        self.Maxpool2 = nn.MaxPool2d(2)                                 # (32,16,8,8)
        self.conv3 = nn.Sequential(nn.Conv2d(16, 32,3, padding=1),      # (32,32,8,8)
                                    nn.BatchNorm2d(32),
                                    nn.ReLU()
                                    )
        self.Maxpool3 = nn.MaxPool2d(2)                                 # 8(32,32,4,4)
        self.fc = nn.Linear(32*4*4, 10)    

    def forward(self, x):
        x = self.conv1(x)
        x = self.Maxpool1(x)
        x = self.conv2(x)
        x = self.Maxpool2(x)
        x = self.conv3(x)
        x = self.Maxpool3(x)
        x = torch.flatten(x, start_dim=1)
        x = self.fc(x)
        return x

# 은닉층 개수 늘리기
class CNN_deep(nn.Module):                                               # (32,3,32,32)
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.conv_block1 = nn.Sequential(nn.Conv2d(3,32,3, padding=1),      # (32,)
                                        nn.BatchNorm2d(32),
                                        nn.ReLU(),
                                        nn.Conv2d(32, 32,3,padding=1),
                                        nn.BatchNorm2d(32),
                                        nn.ReLU()
                                        )
        self.Maxpool1 = nn.MaxPool2d(2) 

        self.conv_block2 = nn.Sequential(nn.Conv2d(32,64,3, padding=1),
                                        nn.BatchNorm2d(64),
                                        nn.ReLU(),
                                        nn.Conv2d(64,64,3,padding=1),
                                        nn.BatchNorm2d(64),
                                        nn.ReLU(),
                                        nn.Conv2d(64,64,3,padding=1) ,
                                        nn.BatchNorm2d(64),
                                        nn.ReLU()
                                        )
        self.Maxpool2 = nn.MaxPool2d(2)

        self.conv_block3 = nn.Sequential(nn.Conv2d(64,128,3, padding=1),
                                        nn.BatchNorm2d(128),
                                        nn.ReLU(),
                                        nn.Conv2d(128,128,3,padding=1),
                                        nn.BatchNorm2d(128),
                                        nn.ReLU(),
                                        nn.Conv2d(128,128,3,padding=1) ,
                                        nn.BatchNorm2d(128),
                                        nn.ReLU()
                                        )
        self.Maxpool3 = nn.MaxPool2d(2)

        self.classifier = nn.Sequential(nn.Linear(128*3*3, 512),
                                        nn.ReLU(),
                                        nn.Linear(512,10))

    def forward(self, x):
        x = self.conv_block1(x)
        x = self.Maxpool1(x)
        x = self.conv_block2(x)
        x = self.Maxpool2(x)   
        x = self.conv_block3(x)
        x = self.Maxpool3(x)   
        x = torch.flatten(x, start_dim=1)                  
        x = self.classifier(x)
        return x
```

```python
model = globals()[model_type]().to(DEVICE)
print(model)

x_batch, _ = next(iter(train_DL))
print(x_batch.shape)
```

```python
if new_model_train:
    optimizer = optim.Adam(model.parameters(), lr=LR)
    loss_history = Train(model, train_DL, criterion, optimizer,EPOCH)

    torch.save(model, save_model_path)

    plt.figure()
    plt.plot(range(1, EPOCH+1), loss_history)
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Train Loss')
    plt.grid()
```

```python
load_model = torch.load(save_model_path, map_location = DEVICE, weights_only = False)
```

```python
Test(load_model, test_DL)
```

```python
load_model.eval()
with torch.no_grad():
    x_batch, y_batch = next(iter(test_DL))
    x_batch = x_batch.to(DEVICE)
    y_batch = y_batch.to(DEVICE)
    y_hat = load_model(x_batch)
    pred = y_hat.argmax(dim=1)



    if model_type =='CNN':
        feature_map1 = load_model.conv1(x_batch)
        feature_map2 = load_model.conv2(load_model.Maxpool1(feature_map1))
        feature_map3 = load_model.conv3(load_model.Maxpool2(feature_map2))

    elif model_type =='CNN_deep':
        feature_map1 = load_model.conv_block1(x_batch)
        feature_map2 = load_model.conv_block2(load_model.Maxpool1(feature_map1))
        feature_map3 = load_model.conv_block3(load_model.Maxpool2(feature_map2))

x_batch = x_batch.cpu()
feature_map1 = feature_map1.cpu()
feature_map2 = feature_map2.cpu()
feature_map3 = feature_map3.cpu()

plt.figure(figsize=(8,8))
plt.xticks([])
plt.yticks([])
plt.imshow(x_batch[0].permute(1,2,0))

print(feature_map1.shape)
plt.figure(figsize=(32,16))
for idx in range(feature_map1.shape[1]):
    plt.subplot(4,8, idx +1, xticks =[], yticks=[])
    plt.imshow(feature_map1[0,idx])

print(feature_map2.shape)
plt.figure(figsize=(32,16))
for idx in range(feature_map2.shape[1]):
    plt.subplot(8,8, idx +1, xticks =[], yticks=[])
    plt.imshow(feature_map2[0,idx])

print(feature_map3.shape)
plt.figure(figsize=(21,8))
for idx in range(feature_map3.shape[1]):
    plt.subplot(8,16, idx +1, xticks =[], yticks=[])
    plt.imshow(feature_map3[0,idx])

summed_map = feature_map3.sum(dim=1)
plt.figure(figsize=(8,8))
plt.xticks([])
plt.yticks([])
plt.imshow(summed_map[0])

plt.figure(figsize=(8,8))
plt.xticks([])
plt.yticks([])
plt.imshow(x_batch[0].permute(1,2,0))
plt.imshow(summed_map[0], extent=[0,32,32,0], alpha=0.4,cmap='viridis')
pred_item = test_DS.classes[pred[0].item()]
true_item = test_DS.classes[y_batch[0].item()]
plt.title(f'{pred_item} ({true_item})', color = 'g' if pred_item == true_item else 'r')
```
