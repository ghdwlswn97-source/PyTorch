#### 1. 기본 수학 함수

| 함수 | 의미 | 예시 | 설명 |
| :---: | :---: | :---: | :--- |
| torch.abs(a) | 절대값 | abx(-3) -> 3 | 모든 원소의 절대값 반환 |
| torch.sqrt(a) | 제곱근 | sqrt(9) -> 3 | $\sqrt{x}$ |
| torch.square(a) | 제곱 | square(3) -> 9 | $x^2$ |
| torch.pow(a, n) | 거듭제곱| pow(2, 3) -> 8 | $x^n$ |
| torch.exp(a) | 지수함수 | exp(1) -> e | $e^x$ |
| torch.log(a) | 자연로그 | log(e) -> 1 | ln(x) |
| torch.log10(a) | 상용로그 | log10(10) -> 1 | $\log_{10}(x)$ |
| torch.log2(a) | 밑이 2인 로그 | log2(2) -> 1 | $\log_2(x)$ |


```python
a = torch.randn(3, 3)
print(a)

# 기본 수학 함수
print('절댓값 :\n', torch.abs(a))
print('제곱근 :\n', torch.sqrt(torch.abs(a)))
print('제곱 :\n', torch.square(a))
print('거듭제곱 : \n', torch.pow(a, 2))

# 지수함수
print('지수함수 : \n',torch.exp(a))
print('자연로그 :\n',torch.log(torch.abs(a)))       # logex
print('자연로그 :\n',torch.log(torch.exp(torch.tensor(1))))     #loge(e)
print('상용로그 :\n', torch.log10(torch.tensor(10)))
print('밑이2로그 :\n', torch.log2(torch.tensor(2)))
```

#### 3. 최소값 / 최대값
| 함수 | 의미 | 설명 | 
| :---: | :---: | :---: | 
|torch.min(a)|최소값|전체원소 중 가장 작은 값|
|torch.max(a)|최대값|전체원소 중 가장 큰 값|
|torch.argmin(a)|최소값의 인덱스|최소값이 위치한 인덱스 반환|
|torch.argmax(a)|최대값의 인덱스|최대값이 위치한 인덱스 반환|


```python
a = torch.randn(3,4)
print(a)

```

```python
print(a)
print()
print('최댓값 : ',torch.max(a))

# 옵션 dim : 각 차원에서 가장 큰값 반환
print('각 열에서 큰값 :',torch.argmax(a, dim=0))       # 열끼리
print('각 행에서 큰값 :',torch.argmax(a, dim=1))       # 행끼리
print()

# 옵션 keepdims : 최댓값을 반환한 후에도 텐서의 차원을 그대로 유지할 것인지 여부(기본값 : false)
print(torch.max(a, dim = 0, keepdim=True))
print('각 행에서 큰값 :',torch.argmax(a, dim=1, keepdim=True))   
print()

print(torch.min(a))
print(torch.argmin(a, dim=0))
print(torch.argmin(a, dim=1))

print(torch.argmax(a))
print(torch.argmax(a, dim=0))
print(torch.argmax(a, dim=1))
```

#### 4. 통계

#### ▶ torch.nan
- not a number(숫자가 아님) : 수학적으로 정의할 수 없거나 컴퓨터가 정상적인 실수값으로 표현할 수 없는 연산 불능 / 수치적인 오류 상태를 의미
    - 정의되지 않은 수학
        - 연산음수의 로그 계산 : torch.log(torch.tensor(-1.0)) -> nan
        - 0을 0으로 나누기 : torch.tensor(0) / torch.tensor(0) -> nan
        - 무한대 간의 뺼셈 : inf - inf -> nan

    - 딥러닝 학습 중 수치적 불안정성(Gradient Explosion 등) 
        - 그라디언트 폭주 : 가중치 업데이트량이 너무 커져 수치가 표현 범위 초과할 때
        - 언더플로우 / 오버플로우

```python
torch.nan

print(torch.log(torch.tensor(-1)))

print(torch.isnan(torch.tensor([1,2,torch.nan, 3,4])))
print(torch.isinf(torch.tensor([1,2,3,4, torch.inf])))
```

#### ▶ 기타함수
### transpose() : 2개의 차원 교환
- 주로 2차원 행렬의 전치(row<-> colnumn)나, 특정 두 축의 위치만 바꿀 때 사용
- 차원의 개수와 상관없이 정확히 2개 축의 인덱스만 넘겨주어야 됨

### permute() : 모든 차원의 순서를 재배치
- 3차원 이상의 고차원 텐서의 차원 순서를 한번에 뒤섞을 때 사용

### t(), T : 행렬 연산(전치)를 간결하게 쓰기위한 메서드 및 속성
- .t() : 2차원 텐서 전용 전치 함수
- .T : 모든 차원을 역순으로 뒤집은 결과를 반환
    - 2차원 행렬 : 일반적인 전치 행렬(A^T)이 됨
    - 고차원 텐서 : 차원의 순서가 완전히 반대로 뒤집힘 (예) (2,3,4) -> (4,3,2)

```python
a = torch.tensor([1,2,3])
b = torch.tensor([2,2,1])
print(torch.sum(a*b))

a = a.reshape(3,1)
b = b.reshape(3,1)
print(a.transpose(1,0)@b)

print(a)
print(a.permute(1,0)@b)
print(a.t()@b)
print(a.T@b)
print()

c = torch.randn(4,3,6)
print(c)
print()

print(c.permute(1,2,0))
print(c.permute(1,2,0).shape)

print(c.transpose(2,1).shape)
```

### squeeze()
- 크기(차원)가 1인 축을 제거하여 텐서의 차원을 축소할 때 사용
- 데이터의 실제 원소 수나 값은 바꾸지 않고, 당지 불필요하게 감싸져 있는 차원(1)만 골라서 차원의 수를 줄여주는 역할

### unsqueeze()
- 지정한 위치에 크기가 1인 차원을 새롭게 추가하여 텐서의 차원을 확장할때 사용
- 실제 데이터나 원소 개수는 그대로 유지되며, 당지 차원의 모양만 변경

```python
a = torch.randn(1,1,1,3,1,1,1,4,1,1,1)
# print(a)
print(a.shape)
print(a.squeeze())
print(a.squeeze().shape)
print(a.squeeze(dim=(0,2,4,5)).shape)
print(a.squeeze(dim=(0,2,4,5,8,9)).shape)

# 축소하고자하는 자리에 값이 있으면 축소하지 않음. 그대로 둠
print(a.squeeze(dim=(0,3,4,5)).shape)
```

```python
a = torch.randn(3,4)
print(a)
print(a.shape)

print(a.unsqueeze(dim=0).shape)
print(a.unsqueeze(dim=1).shape)
print(a.unsqueeze(dim=2).shape)

print(a.reshape(1,3,4).shape)
print(a.reshape(3,1,4).shape)
```
