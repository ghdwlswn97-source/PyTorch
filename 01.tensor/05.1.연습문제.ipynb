#### 문제 02

```python

```

**요구사항**

2. 각 원소의 제곱을 `square_data`에 저장.

#### 문제 18

학습 시간과 문제 풀이 횟수로 시험 점수를 예측하기 위한 데이터를 준비하려고 함.

**요구사항**
1. `pandas.DataFrame`으로 8명의 학습 시간, 문제 풀이 횟수, 시험 점수 데이터를 생성.
2. 시험 점수를 제외한 입력 데이터 `x`와 시험 점수 `y`를 분리.
3. `.values`를 이용해 NumPy 배열로 변환.
4. `y`를 `reshape(-1, 1)`로 2차원 구조로 변경.
5. 입력과 정답을 `torch.float32` 텐서로 변환.
6. `TensorDataset`을 생성하고 데이터 건수 및 첫 3개 데이터 확인.
7. `DataLoader`를 `batch_size=3`, `shuffle=False`로 생성하고 배치 수와 각 배치 데이터를 출력.
8. `shuffle=True`인 DataLoader도 생성하여 첫 번째 배치 확인.

````python
import pandas as pd
from torch.utils.data import TensorDataset, DataLoader
````

````python
data = pd.DataFrame({
    'study_hours' : [1,2,3,4,5,6,7,8],
    'practice_count' : [2,3,3,5,6,7,8,10],
    'exam_score' : [52,58,63,56,76,56,87,67]
})

x = data.drop('exam_score', axis=1)
y = data.loc[:,'exam_score']
# 1차원으로 추출, 뒤에서 reshape 필요.
# reshape 생략하려면 2차원으로 추출
#y = data.loc[:,['exam_score']]
#y = data['exam_score']
x_np = x.values
y_np = y.values.reshape(-1,1)

x_tensor = torch.tensor(x_np, dtype=torch.float32)
y_tensor = torch.tensor(y_np, dtype=torch.float32)

dataset=TensorDataset(x_tensor, y_tensor)

print('데이터 건수 :', len(dataset))
print('첫 3개 데이터 :', dataset[:3])

data_loader = DataLoader(dataset, batch_size =3, shuffle=False)
print('DataLoader 배치수 :', len(data_loader))

for batch_id, (x_b, y_b) in enumerate(data_loader):
    print('Batch ID : ', batch_id)
    print(x_b)
    print(y_b)
    break
````
